```python
X = cc[['tenure', 'monthly_charges', 'total_charges', 'support_calls']]
cc['churn'] = cc['churn'].map({'No': 0, 'Yes': 1})
Y=cc['churn']
```

```python
categorical_cols = ['payment_method','contract','internet_service', 'tech_support', 'online_security']
cc['internet_service'] = cc['internet_service'].fillna('No Service')
df_encoded = pd.get_dummies(cc[categorical_cols], columns=categorical_cols, prefix=categorical_cols, dtype=int)


X = pd.concat([X, df_encoded], axis=1)

print(X.head())
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(X,Y,test_size=0.2,random_state=2)
x_train
y_test
```

```python
print(y_train.value_counts())
```

```python
from imblearn.over_sampling import SMOTE

smote = SMOTE(random_state=42)
x_train_resampled, y_train_resampled = smote.fit_resample(x_train, y_train)
print(x_train_resampled.shape)
print(y_train_resampled.value_counts())
```

```python
from sklearn.preprocessing import StandardScaler

sc=StandardScaler()
x_train_scaled = sc.fit_transform(x_train)
x_test_scaled = sc.transform(x_test)
x_train
```

```python
from sklearn.linear_model import LogisticRegression
lr=LogisticRegression()
lr.fit(x_train_scaled,y_train)
y_pred=lr.predict(x_test_scaled)
y_pred
```

```python
from sklearn import metrics

cnf_matrix = metrics.confusion_matrix(y_test, y_pred)
cnf_matrix
```

```python
from sklearn.metrics import classification_report
target_names = ['will not churn', 'will churn']
print(classification_report(y_test, y_pred, target_names=target_names))
```

```python
print(y_train.value_counts())
```

```python
from sklearn.metrics import roc_auc_score
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(n_estimators=200, random_state=42, n_jobs=-1)
rf.fit(x_train, y_train)
y_prob_rf = rf.predict_proba(x_test)[:, 1]
print(f"Random Forest ROC-AUC: {roc_auc_score(y_test, y_prob_rf):.4f}")
```

```python
import xgboost as xgb
from sklearn.model_selection import cross_val_score

xgb_model = xgb.XGBClassifier()
xgb_model.fit(x_train, y_train)
```

```python

param_grid = {
    'n_estimators': [100, 200, 300],
    'max_depth': [5, 7],
    'learning_rate': [0.05, 0.1],
    'min_child_weight': [3, 5],
}
from sklearn.model_selection import GridSearchCV

xgb_grid = GridSearchCV(
    estimator=xgb.XGBClassifier(random_state=42, tree_method='hist'),
    param_grid=param_grid,
    cv=5,          # 5-fold is more reliable than 3
    n_jobs=-1,
    scoring='roc_auc'  # better than accuracy for churn
)
```

```python
xgb_grid.fit(x_train, y_train)
```

```python
print("Best Parameters:", xgb_grid.best_params_)
print(f"Best CV ROC-AUC: {xgb_grid.best_score_:.4f}")
```

```python
best_model = xgb_grid.best_estimator_
y_pred = best_model.predict(x_test)
y_prob = best_model.predict_proba(x_test)[:, 1]
```

```python
from sklearn.metrics import roc_auc_score
print("\nClassification Report:")
print(classification_report(y_test, y_pred, target_names=['No Churn', 'Churn']))
print(f"Test ROC-AUC: {roc_auc_score(y_test, y_prob):.4f}")
```

```python
xgb_grid.fit(x_train, y_train)
best_model_grid = xgb_grid.best_estimator_

y_pred_grid = best_model_grid.predict(x_test)
y_prob_grid = best_model_grid.predict_proba(x_test)[:, 1]

# Original threshold (0.5)
print("Default threshold (0.5):")
print(classification_report(y_test, y_pred_grid))
print(f"Test ROC-AUC: {roc_auc_score(y_test, y_prob_grid):.4f}")

# Tuned threshold (0.4)
y_pred_tuned = (y_prob_grid >= 0.4).astype(int)
print("Tuned threshold (0.4):")
print(classification_report(y_test, y_pred_tuned))
```

```python
from sklearn.metrics import roc_curve

fpr, tpr, thresholds = roc_curve(y_test, y_prob_grid)
optimal_idx = (tpr - fpr).argmax()
optimal_threshold = thresholds[optimal_idx]
print(f"Optimal threshold: {optimal_threshold:.4f}")

y_pred_optimal = (y_prob_grid >= optimal_threshold).astype(int)
print(classification_report(y_test, y_pred_optimal))
```

```python
from sklearn.metrics import ConfusionMatrixDisplay
ConfusionMatrixDisplay.from_estimator(best_model, x_test, y_test, display_labels=['No Churn', 'Churn'])
plt.title('XGBoost Confusion Matrix')
plt.show()
```

```python
xgb.plot_importance(best_model, max_num_features=10)
plt.title('Top 10 Feature Importances')
plt.show()
```
